```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

```

```python
df_2024_1 = pd.read_csv('Preços semestrais - AUTOMOTIVOS_2024.01.csv', sep = ';', decimal = ',')
df_2024_2 = pd.read_csv('Preços semestrais - AUTOMOTIVOS_2024.02.csv', sep = ';', decimal = ',')
df_2025_1 = pd.read_csv('Preços semestrais - AUTOMOTIVOS_2025.01.csv', sep = ';', decimal = ',')
df_2025_2 = pd.read_csv('Preços semestrais - AUTOMOTIVOS_2025.02.csv', sep = ';', decimal = ',')
```

```python
df = pd.concat([df_2024_1, df_2024_2, df_2025_1, df_2025_2], ignore_index=True)
print(df.shape) 
display(df.head())
```

```python
print(df.columns)
df.head(15)
```

```python
df_gasolinac_rio = df.loc[(df["Estado - Sigla"] == 'RJ') & (df["Produto"] == "GASOLINA")]
df_rio = df_gasolinac_rio.groupby("Municipio")['Valor de Venda'].mean().reset_index().sort_values(by ='Valor de Venda', ascending=False)
df_rio
```

```python
array_gasolina_rio = np.array(df_rio['Valor de Venda'])
media = np.mean(array_gasolina_rio)
mediana = np.median(array_gasolina_rio)
dist = (media - mediana)/mediana 
q1 = np.percentile(array_gasolina_rio,25)
q3= np.percentile(array_gasolina_rio,75)
iqr = q3 - q1
limite_superior = q3 + (1.5*iqr)
limite_inferior = q1 - (1.5*iqr)
print(f"media : {media}")
print(f"mediana : {mediana:.2f}") 
print(f"distancia : {dist:.2%}") 
print(f"Limite Superior : {limite_superior}") 
print(f" Limite inferior = {limite_inferior}") 
print(f"IQR: {iqr}")
```

```python
outlier_rio= df_rio.loc[df_rio['Valor de Venda'] > limite_inferior]
outlier_rio
```

```python
outlier_rio= df_comp.loc[df_comp['Valor de Venda'] < limite_inferior]
outlier_rio
```
