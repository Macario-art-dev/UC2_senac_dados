```python
import pandas as pd
```

```python
dados = "https://www.ispdados.rj.gov.br/Arquivos/BaseDPEvolucaoMensalCisp.csv"
df = pd.read_csv(dados, encoding="latin1", sep=";") 
df
```

```python
# Agrupando por CISP e somando os furtos de celular
df_cisp_roubo = df.groupby('cisp')['roubo_veiculo'].sum().reset_index().head(15) #dentro do parenteses variaveia qualitatias e colchetes as quantitativas 
# Ordenando do maior para o menor
df_cisp_roubo = df_cisp_roubo.sort_values(by='roubo_veiculo', ascending=False)


df_cisp_roubo
```

```python
 #Agrupando por CISP e somando os furtos de celular
df_cisp = df.groupby('cisp')['furto_celular'].sum().reset_index() #dentro do parenteses variaveia qualitatias e colchetes as quantitativas 
# Ordenando do maior para o menor
df_cisp = df_cisp.sort_values(by='furto_celular', ascending=False)
df_cisp
```

```python
df_capital = df.loc[df["munic"]== "Rio de Janeiro"] 
df_cap = df_capital.groupby('cisp')['furto_veiculos'].sum().reset_index() #dentro do parenteses variaveis qualitativas e colchetes as quantitativas 
# Ordenando do maior para o menor
df_cap = df_cap.sort_values(by='furto_veiculos',ascending=False )
df_cap.head(10)
```
